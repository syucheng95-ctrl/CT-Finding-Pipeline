## 用你自己的数据

把 CT 上传到 Google Drive，然后：
```python
# 修改 config.yaml
import yaml
with open('config.yaml') as f:
    cfg = yaml.safe_load(f)
cfg['data']['ct_images'] = '/content/drive/MyDrive/my_ct_data/'
cfg['data']['labels'] = '/content/drive/MyDrive/my_labels/'
with open('config.yaml', 'w') as f:
    yaml.dump(cfg, f)
```
